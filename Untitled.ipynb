```python
import random, numpy as np, pandas as pd, matplotlib.pyplot as plt
import sklearn, xgboost, optuna, shap

SEED, N_SPLITS, TEST_SIZE = 42, 5, 0.20
random.seed(SEED); np.random.seed(SEED)

for lib in (pd, np, sklearn, xgboost, optuna, shap):
    print(lib.__name__, lib.__version__)
```

Partie 2 : EDA orientée décision


```python
df = pd.read_csv("data/diabetic_data.csv", na_values="?")   # "?" -> NaN

print("Forme :", df.shape)
print("\nCible brute :\n", df["readmitted"].value_counts(normalize=True).round(3))
print("\n% manquants :\n",
      (df.isna().mean() * 100).round(1).loc[lambda s: s > 0].sort_values(ascending=False))
print("\nSéjours :", len(df), "| Patients uniques :", df["patient_nbr"].nunique())
print("\nGenre :\n", df["gender"].value_counts())
print("\nSorties décès/hospice :",
      df["discharge_disposition_id"].isin([11, 13, 14, 19, 20, 21]).sum())
print("\nColonnes constantes :", df.nunique()[lambda s: s == 1].index.tolist())
```

```python
df["y"] = (df["readmitted"] == "<30").astype(int)
rate = df.groupby("number_inpatient")["y"].agg(["mean", "count"])
rate = rate[rate["count"] >= 30]            # on ignore les modalités trop rares

rate["mean"].plot(marker="o", figsize=(6, 3.5))
plt.xlabel("number_inpatient"); plt.ylabel("taux de réadmission < 30 j")
plt.title("Réadmission selon les hospitalisations de l'année précédente")
plt.show()
```

Partie 3 : Nettoyage et features

```python
# 3.1 Population exclusions
d = df[~df["discharge_disposition_id"].isin([11, 13, 14, 19, 20, 21])].copy()
d = d[d["gender"] != "Unknown/Invalid"].copy()

# 3.2 ICD-9 grouping (~700 codes -> families)
def icd9_group(code):
    if pd.isna(code): return "Missing"
    c = str(code)
    if c.startswith(("V", "E")): return "Other"
    n = float(c)
    if 250 <= n < 251:                return "Diabetes"
    if 390 <= n <= 459 or n == 785:   return "Circulatory"
    if 460 <= n <= 519 or n == 786:   return "Respiratory"
    if 520 <= n <= 579 or n == 787:   return "Digestive"
    if 580 <= n <= 629 or n == 788:   return "Genitourinary"
    if 710 <= n <= 739:               return "Musculoskeletal"
    if 800 <= n <= 999:               return "Injury"
    if 140 <= n <= 239:               return "Neoplasms"
    return "Other"

for c in ["diag_1", "diag_2", "diag_3"]:
    d[c] = d[c].map(icd9_group)

# 3.3 Domain features
MEDS = ['metformin','repaglinide','nateglinide','chlorpropamide','glimepiride',
        'acetohexamide','glipizide','glyburide','tolbutamide','pioglitazone',
        'rosiglitazone','acarbose','miglitol','troglitazone','tolazamide',
        'examide','citoglipton','insulin','glyburide-metformin','glipizide-metformin',
        'glimepiride-pioglitazone','metformin-rosiglitazone','metformin-pioglitazone']

d["age_mid"]        = d["age"].str.extract(r"\[(\d+)-")[0].astype(int) + 5
d["n_meds_active"]  = (d[MEDS] != "No").sum(axis=1)
d["n_meds_changed"] = d[MEDS].isin(["Up", "Down"]).sum(axis=1)
d["prior_visits"]   = d["number_outpatient"] + d["number_emergency"] + d["number_inpatient"]

d["medical_specialty"] = d["medical_specialty"].fillna("Missing")
freq = d["medical_specialty"].value_counts(normalize=True)
rare = freq[freq < 0.01].index
d["medical_specialty"] = d["medical_specialty"].replace(dict.fromkeys(rare, "Other"))

# 3.4 Final selection and typing
cat_cols = ["race", "gender", "admission_type_id", "discharge_disposition_id",
            "admission_source_id", "medical_specialty", "diag_1", "diag_2", "diag_3",
            "max_glu_serum", "A1Cresult", "change", "diabetesMed"]
num_cols = ["age_mid", "time_in_hospital", "num_lab_procedures", "num_procedures",
            "num_medications", "number_outpatient", "number_emergency", "number_inpatient",
            "number_diagnoses", "n_meds_active", "n_meds_changed", "prior_visits"]

d["race"] = d["race"].fillna("Missing")
for c in ["max_glu_serum", "A1Cresult"]:
    d[c] = d[c].fillna("NotMeasured")
for c in cat_cols:
    d[c] = d[c].astype(str).astype("category")

X      = d[cat_cols + num_cols].reset_index(drop=True)
y      = d["y"].reset_index(drop=True)
groups = d["patient_nbr"].reset_index(drop=True)

print(X.shape, "| prevalence:", round(y.mean(), 4), "| patients:", groups.nunique())
```

Part 4: Split by patient

```python
from sklearn.model_selection import StratifiedGroupKFold

# First fold of a 5-fold grouped split = 20 % of patients as test
sgkf = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEED)
train_idx, test_idx = next(sgkf.split(X, y, groups))

X_train, y_train, g_train = (X.iloc[train_idx].reset_index(drop=True),
                             y.iloc[train_idx].reset_index(drop=True),
                             groups.iloc[train_idx].reset_index(drop=True))
X_test,  y_test,  g_test  = (X.iloc[test_idx].reset_index(drop=True),
                             y.iloc[test_idx].reset_index(drop=True),
                             groups.iloc[test_idx].reset_index(drop=True))

# Sanity checks
assert len(set(g_train) & set(g_test)) == 0, "Patient overlap!"
print("Train:", X_train.shape, "prev", round(y_train.mean(), 4))
print("Test :", X_test.shape,  "prev", round(y_test.mean(), 4))

# Reference CV folds on the train set only (grouped + stratified)
cv = list(StratifiedGroupKFold(n_splits=N_SPLITS, shuffle=True,
                               random_state=SEED).split(X_train, y_train, g_train))
```

Part 5: Baselines

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import cross_validate

scoring = {"roc_auc": "roc_auc", "pr_auc": "average_precision", "brier": "neg_brier_score"}

pre = ColumnTransformer([
    ("num", StandardScaler(), num_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
])

models = {
    "dummy": DummyClassifier(strategy="prior"),
    "logreg": Pipeline([("pre", pre), ("clf", LogisticRegression(max_iter=1000))]),
}

rows = []
for name, m in models.items():
    r = cross_validate(m, X_train, y_train, cv=cv, scoring=scoring, n_jobs=-1)
    rows.append({
        "model": name,
        "ROC-AUC": f"{r['test_roc_auc'].mean():.3f} ± {r['test_roc_auc'].std():.3f}",
        "PR-AUC":  f"{r['test_pr_auc'].mean():.3f} ± {r['test_pr_auc'].std():.3f}",
        "Brier":   f"{-r['test_brier'].mean():.4f} ± {r['test_brier'].std():.4f}",
    })
pd.DataFrame(rows)
```

```python
from sklearn.model_selection import StratifiedKFold
cv_random = list(StratifiedKFold(5, shuffle=True, random_state=SEED).split(X_train, y_train))
r = cross_validate(models["logreg"], X_train, y_train, cv=cv_random, scoring=scoring, n_jobs=-1)
print("Random CV  ROC-AUC:", r["test_roc_auc"].mean().round(4), "PR-AUC:", r["test_pr_auc"].mean().round(4))
```

Part 6: XGBoost

```python
from xgboost import XGBClassifier
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import roc_auc_score, average_precision_score, brier_score_loss

BASE = dict(max_depth=4, min_child_weight=5, subsample=0.8, colsample_bytree=0.8,
            reg_lambda=1.0, gamma=0.0)

def make_xgb(params, **extra):
    return XGBClassifier(
        n_estimators=2000, learning_rate=0.05, early_stopping_rounds=50,
        eval_metric="aucpr", enable_categorical=True, tree_method="hist",
        random_state=SEED, n_jobs=-1, **{**params, **extra})

def cv_xgb(params, X, y, groups, cv):
    oof = np.zeros(len(y)); rows = []; best_iters = []
    for tr, va in cv:
        # Early stopping on 10 % of the training patients (never on the evaluation fold)
        inner = GroupShuffleSplit(n_splits=1, test_size=0.10, random_state=SEED)
        i_fit, i_es = next(inner.split(tr, groups=groups.iloc[tr].values))
        tr_fit, tr_es = tr[i_fit], tr[i_es]

        model = make_xgb(params)
        model.fit(X.iloc[tr_fit], y.iloc[tr_fit],
                  eval_set=[(X.iloc[tr_es], y.iloc[tr_es])], verbose=False)
        p = model.predict_proba(X.iloc[va])[:, 1]
        oof[va] = p
        best_iters.append(model.best_iteration)
        rows.append({"roc_auc": roc_auc_score(y.iloc[va], p),
                     "pr_auc": average_precision_score(y.iloc[va], p),
                     "brier": brier_score_loss(y.iloc[va], p),
                     "mean_risk": p.mean()})
    res = pd.DataFrame(rows)
    print(res.agg(["mean", "std"]).round(4))
    return oof, res, best_iters

oof_v1, res_v1, iters_v1 = cv_xgb(BASE, X_train, y_train, g_train, cv)
print("Best iterations per fold:", iters_v1)
```

```python
tr, va = cv[0]
m = XGBClassifier(n_estimators=400, learning_rate=0.05, eval_metric="aucpr",
                  enable_categorical=True, tree_method="hist", random_state=SEED,
                  n_jobs=-1, **BASE)
m.fit(X_train.iloc[tr], y_train.iloc[tr],
      eval_set=[(X_train.iloc[tr], y_train.iloc[tr]), (X_train.iloc[va], y_train.iloc[va])],
      verbose=False)
ev = m.evals_result()
plt.plot(ev["validation_0"]["aucpr"], label="train")
plt.plot(ev["validation_1"]["aucpr"], label="validation")
plt.xlabel("n trees"); plt.ylabel("PR-AUC"); plt.legend(); plt.show()
print("Best validation iteration:", int(np.argmax(ev["validation_1"]["aucpr"])))
```

Part 7: Class imbalance

```python
spw = (y_train == 0).sum() / (y_train == 1).sum()
print("scale_pos_weight =", round(spw, 2))

oof_w, res_w, iters_w = cv_xgb({**BASE, "scale_pos_weight": spw}, X_train, y_train, g_train, cv)

pd.DataFrame({"v1 (no weight)": res_v1.mean(), "weighted": res_w.mean()}).round(4)
```

Partie 8 : Optimisation des hyperparamètres avec Optuna

```python
import optuna
optuna.logging.set_verbosity(optuna.logging.WARNING)

def objective(trial):
    params = {
        "max_depth":        trial.suggest_int("max_depth", 2, 8),
        "min_child_weight": trial.suggest_float("min_child_weight", 1, 50, log=True),
        "gamma":            trial.suggest_float("gamma", 1e-3, 5, log=True),
        "reg_lambda":       trial.suggest_float("reg_lambda", 1e-2, 50, log=True),
        "reg_alpha":        trial.suggest_float("reg_alpha", 1e-3, 10, log=True),
        "subsample":        trial.suggest_float("subsample", 0.5, 1.0),
        "colsample_bytree": trial.suggest_float("colsample_bytree", 0.4, 1.0),
    }
    scores = []
    for tr, va in cv:
        inner = GroupShuffleSplit(n_splits=1, test_size=0.10, random_state=SEED)
        i_fit, i_es = next(inner.split(tr, groups=g_train.iloc[tr].values))
        m = make_xgb(params)
        m.fit(X_train.iloc[tr[i_fit]], y_train.iloc[tr[i_fit]],
              eval_set=[(X_train.iloc[tr[i_es]], y_train.iloc[tr[i_es]])], verbose=False)
        p = m.predict_proba(X_train.iloc[va])[:, 1]
        scores.append(average_precision_score(y_train.iloc[va], p))
    return np.mean(scores)

study = optuna.create_study(direction="maximize",
                            sampler=optuna.samplers.TPESampler(seed=SEED))
study.optimize(objective, n_trials=40)

print("Meilleure PR-AUC (biaisée vers le haut) :", round(study.best_value, 4))
print(study.best_params)
```

```python
optuna.visualization.matplotlib.plot_param_importances(study)
plt.show()
```

```python
#Ré-évaluation et prédictions OOF

best_params = study.best_params
oof_v2, res_v2, iters_v2 = cv_xgb(best_params, X_train, y_train, g_train, cv)

comp = pd.DataFrame({"xgb_v1": res_v1.mean(), "xgb_optuna": res_v2.mean()}).round(4)
comp.loc["pr_auc_std"] = [res_v1["pr_auc"].std().round(4), res_v2["pr_auc"].std().round(4)]
comp
```

```python
#Entraînement du modèle final sur tout le train
n_final = int(np.mean(iters_v2) * 1.1) + 1

final_model = XGBClassifier(
    n_estimators=n_final, learning_rate=0.05, eval_metric="aucpr",
    enable_categorical=True, tree_method="hist",
    random_state=SEED, n_jobs=-1, **best_params)
final_model.fit(X_train, y_train)
print("Nombre d'arbres du modèle final :", n_final)
```

Partie 9 : Évaluation finale

```python
#Calibration isotonique sur les OOF
from sklearn.isotonic import IsotonicRegression

iso = IsotonicRegression(out_of_bounds="clip", y_min=0, y_max=1)
iso.fit(oof_v2, y_train)                       # appris sur le TRAIN (OOF) uniquement

print("Brier OOF avant calibration :", round(brier_score_loss(y_train, oof_v2), 5))
print("Brier OOF après calibration:", round(brier_score_loss(y_train, iso.predict(oof_v2)), 5))
```

```python
#Choix du seuil sur le train
from sklearn.metrics import precision_recall_curve

p_oof_cal = iso.predict(oof_v2)
prec, rec, thr = precision_recall_curve(y_train, p_oof_cal)
f2 = (5 * prec * rec) / (4 * prec + rec + 1e-12)
threshold = thr[np.argmax(f2[:-1])]
print("Seuil retenu (F2 sur OOF) :", round(float(threshold), 4))
```

```python
#Performance sur le test avec IC à 95 %
# Prédictions test : modèle final + calibrateur
p_test = iso.predict(final_model.predict_proba(X_test)[:, 1])
yt = y_test.values

def metrics_at(y_true, p, thr):
    pred = (p >= thr).astype(int)
    tp = ((pred == 1) & (y_true == 1)).sum(); fn = ((pred == 0) & (y_true == 1)).sum()
    fp = ((pred == 1) & (y_true == 0)).sum(); tn = ((pred == 0) & (y_true == 0)).sum()
    return {"roc_auc": roc_auc_score(y_true, p),
            "pr_auc": average_precision_score(y_true, p),
            "brier": brier_score_loss(y_true, p),
            "sensibilité": tp / (tp + fn), "spécificité": tn / (tn + fp),
            "VPP": tp / (tp + fp) if (tp + fp) else np.nan}

point = metrics_at(yt, p_test, threshold)

# Bootstrap par patient
rng = np.random.default_rng(SEED)
pat_idx = g_test.groupby(g_test).indices            # patient -> positions
patients = np.array(list(pat_idx.keys()))
boots = []
for _ in range(300):
    sample = rng.choice(patients, size=len(patients), replace=True)
    idx = np.concatenate([pat_idx[p] for p in sample])
    boots.append(metrics_at(yt[idx], p_test[idx], threshold))
boots = pd.DataFrame(boots)

table = pd.DataFrame({
    "valeur": pd.Series(point),
    "IC95 bas": boots.quantile(0.025),
    "IC95 haut": boots.quantile(0.975),
}).round(3)
table
```

```python
#Courbes ROC, précision-rappel, calibration
from sklearn.metrics import roc_curve
from sklearn.calibration import calibration_curve

fig, ax = plt.subplots(1, 3, figsize=(15, 4))

fpr, tpr, _ = roc_curve(yt, p_test)
ax[0].plot(fpr, tpr); ax[0].plot([0, 1], [0, 1], "--", c="gray")
ax[0].set(title="ROC", xlabel="1 - spécificité", ylabel="sensibilité")

pr, rc, _ = precision_recall_curve(yt, p_test)
ax[1].plot(rc, pr); ax[1].axhline(yt.mean(), ls="--", c="gray")
ax[1].set(title="Précision-rappel", xlabel="rappel", ylabel="précision")

frac, mean_pred = calibration_curve(yt, p_test, n_bins=10, strategy="quantile")
ax[2].plot(mean_pred, frac, marker="o"); ax[2].plot([0, 0.4], [0, 0.4], "--", c="gray")
ax[2].set(title="Calibration", xlabel="risque prédit", ylabel="fréquence observée")
plt.tight_layout(); plt.show()
```

```python
#Decision curve analysis
def net_benefit(y_true, p, pt):
    n = len(y_true)
    pred = p >= pt
    tp = (pred & (y_true == 1)).sum(); fp = (pred & (y_true == 0)).sum()
    return tp / n - fp / n * pt / (1 - pt)

pts = np.arange(0.02, 0.40, 0.01)
nb_model = [net_benefit(yt, p_test, t) for t in pts]
prev = yt.mean()
nb_all = [prev - (1 - prev) * t / (1 - t) for t in pts]

plt.figure(figsize=(6, 4))
plt.plot(pts, nb_model, label="modèle")
plt.plot(pts, nb_all, label="suivre tout le monde")
plt.axhline(0, c="gray", label="suivre personne")
plt.ylim(-0.02, 0.06); plt.xlabel("seuil p_t"); plt.ylabel("bénéfice net")
plt.legend(); plt.show()

better = [t for t, a, b in zip(pts, nb_model, nb_all) if a > max(b, 0)]
print("Plage où le modèle est supérieur :", round(min(better), 2), "à", round(max(better), 2))
```

Partie 10

```python
#Équité par sous-groupes
age_band = pd.cut(X_test["age_mid"], bins=[0, 40, 60, 80, 100],
                  labels=["<40", "40-59", "60-79", "80+"], right=False)

def audit(groupe, nom):
    rows = []
    for g in groupe.astype(str).unique():
        m = (groupe.astype(str) == g).values
        if m.sum() < 200 or yt[m].sum() < 20:      # effectifs trop faibles : non fiable
            continue
        pred = p_test[m] >= threshold
        rows.append({
            nom: g, "n": m.sum(),
            "prévalence": yt[m].mean(),
            "risque moyen prédit": p_test[m].mean(),
            "ROC-AUC": roc_auc_score(yt[m], p_test[m]),
            "sensibilité": (pred & (yt[m] == 1)).sum() / (yt[m] == 1).sum(),
        })
    return pd.DataFrame(rows).round(3)

display(audit(X_test["gender"], "sexe"))
display(audit(age_band, "âge"))
display(audit(X_test["race"], "origine"))
```

Partie 11 : Interprétabilité avec SHAP

```python
import xgboost as xgb

#Calcul et vue globale
booster = final_model.get_booster()
dtest = xgb.DMatrix(X_test, enable_categorical=True)
contribs = booster.predict(dtest, pred_contribs=True)   # dernière colonne = valeur de base
shap_vals = contribs[:, :-1]
features = list(X_test.columns)

# Importance SHAP : moyenne des |contributions|
shap_imp = pd.Series(np.abs(shap_vals).mean(axis=0), index=features).sort_values(ascending=False)

# Importance "gain" de XGBoost
gain_imp = pd.Series(booster.get_score(importance_type="gain")).reindex(features).fillna(0)
gain_imp = (gain_imp / gain_imp.sum()).sort_values(ascending=False)

fig, ax = plt.subplots(1, 2, figsize=(13, 5))
shap_imp.head(15)[::-1].plot.barh(ax=ax[0], title="Importance SHAP (moy. |SHAP|)")
gain_imp.head(15)[::-1].plot.barh(ax=ax[1], title="Importance gain XGBoost")
plt.tight_layout(); plt.show()

print("Top 5 SHAP :", shap_imp.head(5).index.tolist())
print("Top 5 gain :", gain_imp.head(5).index.tolist())
```

```python
#Effet d’une variable et explication locale
# Effet de number_inpatient sur le risque (échelle log-odds)
j = features.index("number_inpatient")
plt.figure(figsize=(6, 4))
plt.scatter(X_test["number_inpatient"], shap_vals[:, j], s=4, alpha=0.3)
plt.axhline(0, c="gray")
plt.xlabel("number_inpatient"); plt.ylabel("SHAP (log-odds)")
plt.show()

# Explication locale : le patient au risque prédit le plus élevé
i = int(np.argmax(p_test))
contrib_i = pd.Series(shap_vals[i], index=features)
top = contrib_i.reindex(contrib_i.abs().sort_values(ascending=False).index).head(8)
print("Risque prédit :", round(float(p_test[i]), 3), "| réel :", int(yt[i]))
print(pd.DataFrame({"valeur": X_test.iloc[i][top.index].values, "SHAP": top.values.round(3)},
                   index=top.index))
```

```python
#Dépendance de time_in_hospital
k = features.index("time_in_hospital")
plt.figure(figsize=(6, 4))
plt.scatter(X_test["time_in_hospital"], shap_vals[:, k], s=4, alpha=0.3)
plt.axhline(0, c="gray")
plt.xlabel("time_in_hospital (jours)"); plt.ylabel("SHAP (log-odds)")
plt.show()
```

```python
def predire(df_new):
    model = joblib.load("artefacts/xgb_final.joblib")
    calib = joblib.load("artefacts/calibrateur_isotonique.joblib")
    sch = json.load(open("artefacts/schema.json"))
    X_new = df_new[sch["cat_cols"] + sch["num_cols"]].copy()
    for c in sch["cat_cols"]:
        X_new[c] = pd.Categorical(X_new[c].astype(str), categories=sch["categories"][c])
    p = calib.predict(model.predict_proba(X_new)[:, 1])
    return p, (p >= sch["seuil"]).astype(int)

p_rel, _ = predire(X_test)
print("Prédictions identiques :", np.allclose(p_rel, p_test))
```

# Model card : Prédiction de la réadmission à 30 jours (patients diabétiques)

| Rubrique | Contenu |
|---|---|
| **Usage prévu** | Aide à la décision pour **cibler un suivi renforcé** (appel infirmier, consultation précoce, éducation thérapeutique) après la sortie d'un séjour hospitalier de patient diabétique. Le score assiste le clinicien et ne le remplace pas. |
| **Usages exclus** | Prédiction à l'admission (variables connues seulement à la sortie) ; décision de refuser ou limiter des soins ; usage assurantiel ; utilisation dans un autre pays/hôpital sans revalidation ; usage sur patients non diabétiques ou pédiatriques. |
| **Population et données** | Diabetes 130-US Hospitals (1999-2008). 99 340 séjours, 69 987 patients après exclusion des décès/hospice et genres invalides. Découpage par patient : train 79 473 / test 19 867. Prévalence de la cible : 11,4 %. |
| **Cible et horizon** | `y = 1` si réadmission **< 30 jours**, sinon 0. Prédiction au moment de la sortie. |
